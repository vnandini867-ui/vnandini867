```python
# ============================================================
# CNN ON MNIST - COMPLETE EVALUATION
# Classification Report | ROC-AUC | Accuracy | Loss | Confusion Matrix
# ============================================================

import tensorflow as tf
from tensorflow.keras import layers, models
from tensorflow.keras.callbacks import EarlyStopping
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
from sklearn.metrics import classification_report, confusion_matrix
from sklearn.metrics import roc_curve, auc
from sklearn.preprocessing import label_binarize

# ============================================================
# 1. LOAD MNIST DATASET
# ============================================================

(x_train, y_train), (x_test, y_test) = tf.keras.datasets.mnist.load_data()

print("Training data:", x_train.shape)
print("Testing data :", x_test.shape)

# Normalize pixel values: X' = X / 255
x_train = x_train.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0

# CNN input: height x width x channels
x_train = x_train.reshape(-1, 28, 28, 1)
x_test = x_test.reshape(-1, 28, 28, 1)

# ============================================================
# 2. CREATE CNN MODEL
# ============================================================

model = models.Sequential([
    layers.Conv2D(32, (3,3), activation='relu',
                  input_shape=(28,28,1)),
    layers.MaxPooling2D((2,2)),

    layers.Conv2D(64, (3,3), activation='relu'),
    layers.MaxPooling2D((2,2)),

    layers.Flatten(),
    layers.Dense(128, activation='relu'),
    layers.Dropout(0.5),

    # 10 classes: digits 0-9
    layers.Dense(10, activation='softmax')
])

model.summary()

# ============================================================
# 3. COMPILE MODEL
# ============================================================

model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

# ============================================================
# 4. EARLY STOPPING
# ============================================================

early_stopping = EarlyStopping(
    monitor='val_loss',
    patience=3,
    restore_best_weights=True
)

# ============================================================
# 5. TRAIN MODEL
# ============================================================

history = model.fit(
    x_train,
    y_train,
    epochs=20,
    batch_size=32,
    validation_split=0.1,
    callbacks=[early_stopping],
    verbose=1
)

# ============================================================
# 6. TEST ACCURACY AND LOSS
# ============================================================

test_loss, test_accuracy = model.evaluate(
    x_test, y_test, verbose=0
)

print("\n==========================================")
print("TEST RESULTS")
print("==========================================")
print("Test Loss     :", test_loss)
print("Test Accuracy :", test_accuracy)

# ============================================================
# 7. PREDICTIONS
# ============================================================

y_probability = model.predict(x_test, verbose=0)

# Predicted class
y_pred = np.argmax(y_probability, axis=1)

# ============================================================
# 8. CLASSIFICATION REPORT
# ============================================================

print("\n==========================================")
print("CLASSIFICATION REPORT")
print("==========================================")

print(classification_report(
    y_test,
    y_pred,
    target_names=[str(i) for i in range(10)]
))

# ============================================================
# 9. CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(y_test, y_pred)

plt.figure(figsize=(8,6))
sns.heatmap(
    cm,
    annot=True,
    fmt='d',
    cmap='Blues',
    xticklabels=range(10),
    yticklabels=range(10)
)

plt.title("Confusion Matrix")
plt.xlabel("Predicted Label")
plt.ylabel("True Label")
plt.show()

# ============================================================
# 10. ROC CURVE AND AUC
# ============================================================

# Convert labels to one-hot format for ROC
y_test_binary = label_binarize(y_test, classes=range(10))

plt.figure(figsize=(8,6))

roc_auc = {}

for i in range(10):
    fpr, tpr, _ = roc_curve(
        y_test_binary[:, i],
        y_probability[:, i]
    )

    roc_auc[i] = auc(fpr, tpr)

    plt.plot(
        fpr,
        tpr,
        label=f'Digit {i} (AUC = {roc_auc[i]:.3f})'
    )

plt.plot(
    [0,1], [0,1],
    'k--',
    label='Random Classifier'
)

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve - MNIST")
plt.legend(loc="lower right", fontsize=8)
plt.grid()
plt.show()

print("\n==========================================")
print("ROC-AUC SCORES")
print("==========================================")

for i in range(10):
    print(f"Digit {i}: AUC = {roc_auc[i]:.4f}")

# Macro-average AUC
macro_auc = np.mean(list(roc_auc.values()))
print("\nMacro Average AUC:", round(macro_auc, 4))

# ============================================================
# 11. TRAINING & VALIDATION ACCURACY
# ============================================================

plt.figure(figsize=(8,5))

plt.plot(
    history.history['accuracy'],
    label='Training Accuracy'
)

plt.plot(
    history.history['val_accuracy'],
    label='Validation Accuracy'
)

plt.title("Training and Validation Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()
plt.grid()
plt.show()

# ============================================================
# 12. TRAINING & VALIDATION LOSS
# ============================================================

plt.figure(figsize=(8,5))

plt.plot(
    history.history['loss'],
    label='Training Loss'
)

plt.plot(
    history.history['val_loss'],
    label='Validation Loss'
)

plt.title("Training and Validation Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.grid()
plt.show()

# ============================================================
# 13. SAMPLE PREDICTION
# ============================================================

index = 0

prediction = model.predict(
    x_test[index:index+1],
    verbose=0
)

predicted_digit = np.argmax(prediction)

print("\n==========================================")
print("SAMPLE PREDICTION")
print("==========================================")
print("Predicted Digit:", predicted_digit)
print("Actual Digit   :", y_test[index])

plt.imshow(
    x_test[index].reshape(28,28),
    cmap='gray'
)

plt.title(
    f"Predicted: {predicted_digit} | Actual: {y_test[index]}"
)

plt.axis('off')
plt.show()
```

```python
# ============================================================
# CAT VS DOG CNN - GOOGLE COLAB
# COMPLETE CORRECTED VERSION
# ============================================================

# ------------------------------------------------------------
# 1. INSTALL REQUIRED LIBRARIES
# ------------------------------------------------------------



# ------------------------------------------------------------
# 2. IMPORT LIBRARIES
# ------------------------------------------------------------

import tensorflow as tf
import tensorflow_datasets as tfds

import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from tensorflow.keras import layers, models
from tensorflow.keras.callbacks import EarlyStopping
from tensorflow.keras.utils import load_img, img_to_array

from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    roc_curve,
    roc_auc_score
)

print("TensorFlow version:", tf.__version__)


# ------------------------------------------------------------
# 3. LOAD CAT VS DOG DATASET
# ------------------------------------------------------------

print("\nLoading Cat vs Dog dataset...")

dataset, info = tfds.load(
    "cats_vs_dogs",
    split="train",
    as_supervised=True,
    with_info=True
)

total_images = info.splits["train"].num_examples

print("Dataset loaded successfully!")
print("Total images:", total_images)


# ------------------------------------------------------------
# 4. SPLIT DATASET
# ------------------------------------------------------------

dataset = dataset.shuffle(
    total_images,
    reshuffle_each_iteration=False
)

train_size = int(total_images * 0.8)
validation_size = int(total_images * 0.1)
test_size = total_images - train_size - validation_size

train_dataset = dataset.take(train_size)

remaining_dataset = dataset.skip(train_size)

validation_dataset = remaining_dataset.take(
    validation_size
)

test_dataset = remaining_dataset.skip(
    validation_size
)

print("\n========================================")
print("DATASET SPLIT")
print("========================================")

print("Training images   :", train_size)
print("Validation images :", validation_size)
print("Testing images    :", test_size)


# ------------------------------------------------------------
# 5. IMAGE SIZE AND BATCH SIZE
# ------------------------------------------------------------

IMAGE_SIZE = (150, 150)

BATCH_SIZE = 32


# ------------------------------------------------------------
# 6. PREPROCESSING
# ------------------------------------------------------------

def preprocess(image, label):

    # Resize image
    image = tf.image.resize(
        image,
        IMAGE_SIZE
    )

    # Convert pixel values to float
    image = tf.cast(
        image,
        tf.float32
    )

    # Normalize pixels
    image = image / 255.0

    return image, label


# ------------------------------------------------------------
# 7. DATA AUGMENTATION
# ------------------------------------------------------------

data_augmentation = tf.keras.Sequential([

    layers.RandomFlip(
        "horizontal"
    ),

    layers.RandomRotation(
        0.2
    ),

    layers.RandomZoom(
        0.2
    ),

    layers.RandomTranslation(
        0.2,
        0.2
    )
])


# ------------------------------------------------------------
# 8. PREPARE TRAINING DATA
# ------------------------------------------------------------

train_dataset = train_dataset.map(
    preprocess,
    num_parallel_calls=tf.data.AUTOTUNE
)

train_dataset = train_dataset.batch(
    BATCH_SIZE
)

train_dataset = train_dataset.prefetch(
    tf.data.AUTOTUNE
)


# ------------------------------------------------------------
# 9. PREPARE VALIDATION DATA
# ------------------------------------------------------------

validation_dataset = validation_dataset.map(
    preprocess,
    num_parallel_calls=tf.data.AUTOTUNE
)

validation_dataset = validation_dataset.batch(
    BATCH_SIZE
)

validation_dataset = validation_dataset.prefetch(
    tf.data.AUTOTUNE
)


# ------------------------------------------------------------
# 10. PREPARE TEST DATA
# ------------------------------------------------------------

test_dataset = test_dataset.map(
    preprocess,
    num_parallel_calls=tf.data.AUTOTUNE
)

test_dataset = test_dataset.batch(
    BATCH_SIZE
)

test_dataset = test_dataset.prefetch(
    tf.data.AUTOTUNE
)


# ============================================================
# 11. CAT VS DOG VISUALIZATION
# ============================================================

print("\nDisplaying Cat and Dog images...")

images, labels = next(
    iter(train_dataset)
)

plt.figure(
    figsize=(12, 9)
)

for i in range(12):

    plt.subplot(
        3,
        4,
        i + 1
    )

    plt.imshow(
        images[i].numpy()
    )

    # IMPORTANT:
    # TFDS Cats vs Dogs:
    # 0 = Cat
    # 1 = Dog

    if labels[i].numpy() == 0:
        label_name = "CAT"
    else:
        label_name = "DOG"

    plt.title(
        label_name,
        fontsize=14
    )

    plt.axis("off")


plt.suptitle(
    "CAT VS DOG DATASET VISUALIZATION",
    fontsize=18
)

plt.tight_layout()

plt.show()


# ============================================================
# 12. CREATE CNN MODEL
# ============================================================

model = models.Sequential([

    # Input layer
    layers.Input(
        shape=(150, 150, 3)
    ),

    # Data augmentation
    data_augmentation,

    # CNN Layer 1
    layers.Conv2D(
        32,
        (3, 3),
        activation="relu"
    ),

    layers.MaxPooling2D(
        (2, 2)
    ),


    # CNN Layer 2
    layers.Conv2D(
        64,
        (3, 3),
        activation="relu"
    ),

    layers.MaxPooling2D(
        (2, 2)
    ),


    # CNN Layer 3
    layers.Conv2D(
        128,
        (3, 3),
        activation="relu"
    ),

    layers.MaxPooling2D(
        (2, 2)
    ),


    # CNN Layer 4
    layers.Conv2D(
        128,
        (3, 3),
        activation="relu"
    ),

    layers.MaxPooling2D(
        (2, 2)
    ),


    # Convert to 1D
    layers.Flatten(),


    # Fully connected layer
    layers.Dense(
        128,
        activation="relu"
    ),


    # Dropout
    layers.Dropout(
        0.5
    ),


    # OUTPUT
    # 0 = Cat
    # 1 = Dog
    layers.Dense(
        1,
        activation="sigmoid"
    )
])


# ============================================================
# 13. MODEL SUMMARY
# ============================================================

print("\n========================================")
print("CNN MODEL")
print("========================================")

model.summary()


# ============================================================
# 14. COMPILE MODEL
# ============================================================

model.compile(

    optimizer="adam",

    loss="binary_crossentropy",

    metrics=["accuracy"]
)


# ============================================================
# 15. EARLY STOPPING
# ============================================================

early_stopping = EarlyStopping(

    monitor="val_loss",

    patience=5,

    restore_best_weights=True
)


# ============================================================
# 16. TRAIN MODEL
# ============================================================

print("\n========================================")
print("STARTING CNN TRAINING")
print("========================================")

history = model.fit(

    train_dataset,

    validation_data=validation_dataset,

    epochs=30,

    callbacks=[
        early_stopping
    ]
)


# ============================================================
# 17. ACCURACY GRAPH
# ============================================================

plt.figure(
    figsize=(8, 5)
)

plt.plot(
    history.history["accuracy"],
    label="Training Accuracy"
)

plt.plot(
    history.history["val_accuracy"],
    label="Validation Accuracy"
)

plt.title(
    "Cat vs Dog - Accuracy"
)

plt.xlabel(
    "Epoch"
)

plt.ylabel(
    "Accuracy"
)

plt.legend()

plt.grid()

plt.show()


# ============================================================
# 18. LOSS GRAPH
# ============================================================

plt.figure(
    figsize=(8, 5)
)

plt.plot(
    history.history["loss"],
    label="Training Loss"
)

plt.plot(
    history.history["val_loss"],
    label="Validation Loss"
)

plt.title(
    "Cat vs Dog - Loss"
)

plt.xlabel(
    "Epoch"
)

plt.ylabel(
    "Loss"
)

plt.legend()

plt.grid()

plt.show()


# ============================================================
# 19. TEST MODEL
# ============================================================

test_loss, test_accuracy = model.evaluate(
    test_dataset
)

print("\n========================================")
print("TEST RESULTS")
print("========================================")

print(
    "Test Loss:",
    round(test_loss, 4)
)

print(
    "Test Accuracy:",
    round(
        test_accuracy * 100,
        2
    ),
    "%"
)


# ============================================================
# 20. GET PREDICTIONS
# ============================================================

actual_labels = []

predicted_probabilities = []


for images, labels in test_dataset:

    predictions = model.predict(
        images,
        verbose=0
    )

    actual_labels.extend(
        labels.numpy()
    )

    predicted_probabilities.extend(
        predictions.flatten()
    )


actual_labels = np.array(
    actual_labels
)

predicted_probabilities = np.array(
    predicted_probabilities
)


# Probability >= 0.5 = Dog
# Probability < 0.5 = Cat

predicted_labels = (
    predicted_probabilities >= 0.5
).astype(int)


# ============================================================
# 21. CLASSIFICATION REPORT
# ============================================================

print("\n========================================")
print("CLASSIFICATION REPORT")
print("========================================")

print(
    classification_report(
        actual_labels,
        predicted_labels,
        target_names=[
            "Cat",
            "Dog"
        ]
    )
)


# ============================================================
# 22. CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(
    actual_labels,
    predicted_labels
)

plt.figure(
    figsize=(7, 6)
)

sns.heatmap(

    cm,

    annot=True,

    fmt="d",

    cmap="Blues",

    xticklabels=[
        "Cat",
        "Dog"
    ],

    yticklabels=[
        "Cat",
        "Dog"
    ]
)

plt.title(
    "Cat vs Dog - Confusion Matrix"
)

plt.xlabel(
    "Predicted"
)

plt.ylabel(
    "Actual"
)

plt.show()


# ============================================================
# 23. ROC CURVE
# ============================================================

fpr, tpr, thresholds = roc_curve(

    actual_labels,

    predicted_probabilities
)


auc_score = roc_auc_score(

    actual_labels,

    predicted_probabilities
)


print("\n========================================")
print("ROC-AUC RESULT")
print("========================================")

print(
    "AUC Score:",
    round(
        auc_score,
        4
    )
)


plt.figure(
    figsize=(8, 6)
)

plt.plot(

    fpr,

    tpr,

    label="CNN (AUC = {:.4f})".format(
        auc_score
    )
)

plt.plot(

    [0, 1],

    [0, 1],

    linestyle="--",

    label="Random Classifier"
)

plt.xlabel(
    "False Positive Rate"
)

plt.ylabel(
    "True Positive Rate"
)

plt.title(
    "ROC Curve - Cat vs Dog"
)

plt.legend()

plt.grid()

plt.show()


# ============================================================
# 24. SAVE MODEL
# ============================================================

model.save(
    "/content/cat_dog_CNN_model.keras"
)

print("\n========================================")
print("MODEL SAVED SUCCESSFULLY")
print("========================================")

print(
    "/content/cat_dog_CNN_model.keras"
)


# ============================================================
# 25. UPLOAD YOUR OWN CAT OR DOG IMAGE
# ============================================================

from google.colab import files

print("\n========================================")
print("UPLOAD YOUR CAT OR DOG IMAGE")
print("========================================")

uploaded = files.upload()


# ============================================================
# 26. PREDICT UPLOADED IMAGE
# ============================================================

for filename in uploaded.keys():

    image = load_img(
        filename,
        target_size=IMAGE_SIZE
    )

    image_array = img_to_array(
        image
    )

    image_array = image_array / 255.0

    image_array = np.expand_dims(
        image_array,
        axis=0
    )

    prediction = model.predict(
        image_array,
        verbose=0
    )[0][0]


    # 0 = Cat
    # 1 = Dog

    if prediction >= 0.5:

        predicted_class = "DOG"

        confidence = prediction * 100

    else:

        predicted_class = "CAT"

        confidence = (
            1 - prediction
        ) * 100


    plt.figure(
        figsize=(6, 6)
    )

    plt.imshow(
        image
    )

    plt.title(
        f"Prediction: {predicted_class}\n"
        f"Confidence: {confidence:.2f}%"
    )

    plt.axis("off")

    plt.show()


    print("\n========================================")

    print(
        "Predicted Class:",
        predicted_class
    )

    print(
        "Confidence:",
        round(
            confidence,
            2
        ),
        "%"
    )
```
