```python
# ============================================================
# SKETCH TO REAL IMAGE GENERATION USING Pix2Pix GAN
# Google Colab - Complete Code in One Stretch
# ============================================================

# ------------------------------------------------------------
# 1. IMPORT LIBRARIES
# ------------------------------------------------------------

import os
import glob
import time
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf

from tensorflow.keras import layers

print("TensorFlow version:", tf.__version__)
print("GPU:", tf.config.list_physical_devices('GPU'))


# ------------------------------------------------------------
# 2. DOWNLOAD DATASET
# ------------------------------------------------------------

print("\nDownloading Facades dataset...")

!wget -q http://efrosgans.eecs.berkeley.edu/pix2pix/datasets/facades.tar.gz

print("Download completed.")


# ------------------------------------------------------------
# 3. EXTRACT DATASET
# ------------------------------------------------------------

print("\nExtracting dataset...")

!tar -xzf facades.tar.gz

print("Dataset extracted.")


# ------------------------------------------------------------
# 4. CHECK DATASET
# ------------------------------------------------------------

train_images = glob.glob("facades/train/*.jpg")
val_images = glob.glob("facades/val/*.jpg")
test_images = glob.glob("facades/test/*.jpg")

print("\nDataset information")
print("-------------------")
print("Training images :", len(train_images))
print("Validation images:", len(val_images))
print("Test images      :", len(test_images))


# ------------------------------------------------------------
# 5. PARAMETERS
# ------------------------------------------------------------

IMG_WIDTH = 256
IMG_HEIGHT = 256

BUFFER_SIZE = 400
BATCH_SIZE = 1

OUTPUT_CHANNELS = 3

# Number of training epochs
# Start with 20 for testing.
# Increase to 50 or 100 for better results.
EPOCHS = 20

LAMBDA = 100


# ------------------------------------------------------------
# 6. LOAD IMAGE
# ------------------------------------------------------------

def load_image(image_file):

    image = tf.io.read_file(image_file)

    image = tf.image.decode_jpeg(
        image,
        channels=3
    )

    image = tf.cast(
        image,
        tf.float32
    )

    return image


# ------------------------------------------------------------
# 7. SPLIT IMAGE
# ------------------------------------------------------------

# Facades images contain:
#
#       Input Image | Target Image
#
# We separate them into two images.

def split_image(image):

    width = tf.shape(image)[1]

    half_width = width // 2

    input_image = image[:, :half_width, :]

    target_image = image[:, half_width:, :]

    return input_image, target_image


# ------------------------------------------------------------
# 8. RESIZE IMAGES
# ------------------------------------------------------------

def resize_images(
    input_image,
    target_image
):

    input_image = tf.image.resize(
        input_image,
        [IMG_HEIGHT, IMG_WIDTH]
    )

    target_image = tf.image.resize(
        target_image,
        [IMG_HEIGHT, IMG_WIDTH]
    )

    return input_image, target_image


# ------------------------------------------------------------
# 9. NORMALIZE IMAGES
# ------------------------------------------------------------

# Convert pixel values:
#
# 0 to 255
#
# into:
#
# -1 to +1

def normalize(
    input_image,
    target_image
):

    input_image = (
        input_image / 127.5
    ) - 1

    target_image = (
        target_image / 127.5
    ) - 1

    return input_image, target_image


# ------------------------------------------------------------
# 10. COMPLETE IMAGE PREPROCESSING
# ------------------------------------------------------------

def load_train_image(image_file):

    image = load_image(image_file)

    input_image, target_image = split_image(
        image
    )

    input_image, target_image = resize_images(
        input_image,
        target_image
    )

    input_image, target_image = normalize(
        input_image,
        target_image
    )

    return input_image, target_image


# ------------------------------------------------------------
# 11. CREATE TRAINING DATASET
# ------------------------------------------------------------

train_dataset = tf.data.Dataset.from_tensor_slices(
    train_images
)

train_dataset = train_dataset.map(
    load_train_image,
    num_parallel_calls=tf.data.AUTOTUNE
)

train_dataset = train_dataset.shuffle(
    BUFFER_SIZE
)

train_dataset = train_dataset.batch(
    BATCH_SIZE
)

train_dataset = train_dataset.prefetch(
    tf.data.AUTOTUNE
)


# ------------------------------------------------------------
# 12. CREATE VALIDATION DATASET
# ------------------------------------------------------------

val_dataset = tf.data.Dataset.from_tensor_slices(
    val_images
)

val_dataset = val_dataset.map(
    load_train_image,
    num_parallel_calls=tf.data.AUTOTUNE
)

val_dataset = val_dataset.batch(
    BATCH_SIZE
)

val_dataset = val_dataset.prefetch(
    tf.data.AUTOTUNE
)


# ------------------------------------------------------------
# 13. DISPLAY ORIGINAL DATA
# ------------------------------------------------------------

sample_input, sample_target = next(
    iter(train_dataset)
)

plt.figure(figsize=(12, 6))

plt.subplot(1, 2, 1)

plt.imshow(
    (sample_input[0] + 1) / 2
)

plt.title("Input / Sketch")

plt.axis("off")


plt.subplot(1, 2, 2)

plt.imshow(
    (sample_target[0] + 1) / 2
)

plt.title("Real Image")

plt.axis("off")

plt.show()


# ------------------------------------------------------------
# 14. DOWNSAMPLE BLOCK
# ------------------------------------------------------------

def downsample(
    filters,
    size,
    apply_batchnorm=True
):

    initializer = tf.random_normal_initializer(
        0.,
        0.02
    )

    result = tf.keras.Sequential()

    result.add(
        layers.Conv2D(
            filters,
            size,
            strides=2,
            padding="same",
            kernel_initializer=initializer,
            use_bias=False
        )
    )

    if apply_batchnorm:

        result.add(
            layers.BatchNormalization()
        )

    result.add(
        layers.LeakyReLU()
    )

    return result


# ------------------------------------------------------------
# 15. UPSAMPLE BLOCK
# ------------------------------------------------------------

def upsample(
    filters,
    size,
    apply_dropout=False
):

    initializer = tf.random_normal_initializer(
        0.,
        0.02
    )

    result = tf.keras.Sequential()

    result.add(
        layers.Conv2DTranspose(
            filters,
            size,
            strides=2,
            padding="same",
            kernel_initializer=initializer,
            use_bias=False
        )
    )

    result.add(
        layers.BatchNormalization()
    )

    if apply_dropout:

        result.add(
            layers.Dropout(0.5)
        )

    result.add(
        layers.ReLU()
    )

    return result


# ------------------------------------------------------------
# 16. GENERATOR - U-NET
# ------------------------------------------------------------

def Generator():

    inputs = layers.Input(
        shape=[256, 256, 3]
    )

    # Encoder
    down_stack = [

        downsample(
            64,
            4,
            apply_batchnorm=False
        ),

        downsample(
            128,
            4
        ),

        downsample(
            256,
            4
        ),

        downsample(
            512,
            4
        ),

        downsample(
            512,
            4
        ),

        downsample(
            512,
            4
        ),

        downsample(
            512,
            4
        ),

        downsample(
            512,
            4
        )
    ]


    # Decoder
    up_stack = [

        upsample(
            512,
            4,
            apply_dropout=True
        ),

        upsample(
            512,
            4,
            apply_dropout=True
        ),

        upsample(
            512,
            4,
            apply_dropout=True
        ),

        upsample(
            512,
            4
        ),

        upsample(
            256,
            4
        ),

        upsample(
            128,
            4
        ),

        upsample(
            64,
            4
        )
    ]


    initializer = tf.random_normal_initializer(
        0.,
        0.02
    )


    last = layers.Conv2DTranspose(
        OUTPUT_CHANNELS,
        4,
        strides=2,
        padding="same",
        kernel_initializer=initializer,
        activation="tanh"
    )


    x = inputs

    skips = []


    # Encoder
    for down in down_stack:

        x = down(x)

        skips.append(x)


    # Remove bottleneck
    skips = reversed(
        skips[:-1]
    )


    # Decoder
    for up, skip in zip(
        up_stack,
        skips
    ):

        x = up(x)

        x = layers.Concatenate()(
            [x, skip]
        )


    x = last(x)

    return tf.keras.Model(
        inputs=inputs,
        outputs=x
    )


# ------------------------------------------------------------
# 17. CREATE GENERATOR
# ------------------------------------------------------------

generator = Generator()

print("\nGenerator created successfully.")

generator.summary()


# ------------------------------------------------------------
# 18. TEST GENERATOR BEFORE TRAINING
# ------------------------------------------------------------

generated_image = generator(
    sample_input,
    training=False
)

print(
    "\nGenerated image shape:",
    generated_image.shape
)


# ------------------------------------------------------------
# 19. DISPLAY BEFORE TRAINING
# ------------------------------------------------------------

plt.figure(figsize=(15, 5))


plt.subplot(1, 3, 1)

plt.imshow(
    (sample_input[0] + 1) / 2
)

plt.title("Input Sketch")

plt.axis("off")


plt.subplot(1, 3, 2)

plt.imshow(
    (sample_target[0] + 1) / 2
)

plt.title("Real Image")

plt.axis("off")


plt.subplot(1, 3, 3)

plt.imshow(
    (generated_image[0] + 1) / 2
)

plt.title("Before Training")

plt.axis("off")

plt.show()


# ------------------------------------------------------------
# 20. DISCRIMINATOR - PatchGAN
# ------------------------------------------------------------

def Discriminator():

    initializer = tf.random_normal_initializer(
        0.,
        0.02
    )


    inp = layers.Input(
        shape=[256, 256, 3],
        name="input_image"
    )


    tar = layers.Input(
        shape=[256, 256, 3],
        name="target_image"
    )


    # Combine input and target
    x = layers.concatenate(
        [inp, tar]
    )


    down1 = downsample(
        64,
        4,
        False
    )(x)


    down2 = downsample(
        128,
        4
    )(down1)


    down3 = downsample(
        256,
        4
    )(down2)


    zero_pad1 = layers.ZeroPadding2D()(
        down3
    )


    conv = layers.Conv2D(
        512,
        4,
        strides=1,
        kernel_initializer=initializer,
        use_bias=False
    )(zero_pad1)


    batchnorm1 = layers.BatchNormalization()(
        conv
    )


    leaky_relu = layers.LeakyReLU()(
        batchnorm1
    )


    zero_pad2 = layers.ZeroPadding2D()(
        leaky_relu
    )


    last = layers.Conv2D(
        1,
        4,
        strides=1,
        kernel_initializer=initializer
    )(zero_pad2)


    return tf.keras.Model(
        inputs=[inp, tar],
        outputs=last
    )


# ------------------------------------------------------------
# 21. CREATE DISCRIMINATOR
# ------------------------------------------------------------

discriminator = Discriminator()

print("\nDiscriminator created successfully.")

discriminator.summary()


# ------------------------------------------------------------
# 22. TEST DISCRIMINATOR
# ------------------------------------------------------------

decision = discriminator(
    [
        sample_input,
        sample_target
    ],
    training=False
)

print(
    "\nDiscriminator output shape:",
    decision.shape
)


# ------------------------------------------------------------
# 23. LOSS FUNCTION
# ------------------------------------------------------------

loss_object = tf.keras.losses.BinaryCrossentropy(
    from_logits=True
)


# ------------------------------------------------------------
# 24. GENERATOR LOSS
# ------------------------------------------------------------

def generator_loss(
    disc_generated_output,
    gen_output,
    target
):

    # GAN loss
    gan_loss = loss_object(
        tf.ones_like(
            disc_generated_output
        ),
        disc_generated_output
    )


    # L1 loss
    l1_loss = tf.reduce_mean(
        tf.abs(
            target - gen_output
        )
    )


    # Total generator loss
    total_gen_loss = (
        gan_loss +
        (LAMBDA * l1_loss)
    )


    return (
        total_gen_loss,
        gan_loss,
        l1_loss
    )


# ------------------------------------------------------------
# 25. DISCRIMINATOR LOSS
# ------------------------------------------------------------

def discriminator_loss(
    disc_real_output,
    disc_generated_output
):

    real_loss = loss_object(
        tf.ones_like(
            disc_real_output
        ),
        disc_real_output
    )


    generated_loss = loss_object(
        tf.zeros_like(
            disc_generated_output
        ),
        disc_generated_output
    )


    total_disc_loss = (
        real_loss +
        generated_loss
    )


    return total_disc_loss


# ------------------------------------------------------------
# 26. OPTIMIZERS
# ------------------------------------------------------------

generator_optimizer = tf.keras.optimizers.Adam(
    learning_rate=2e-4,
    beta_1=0.5
)


discriminator_optimizer = tf.keras.optimizers.Adam(
    learning_rate=2e-4,
    beta_1=0.5
)


# ------------------------------------------------------------
# 27. CHECKPOINT
# ------------------------------------------------------------

checkpoint_dir = "./training_checkpoints"

checkpoint_prefix = os.path.join(
    checkpoint_dir,
    "ckpt"
)


checkpoint = tf.train.Checkpoint(
    generator_optimizer=generator_optimizer,
    discriminator_optimizer=discriminator_optimizer,
    generator=generator,
    discriminator=discriminator
)


# ------------------------------------------------------------
# 28. TRAINING STEP
# ------------------------------------------------------------

@tf.function
def train_step(
    input_image,
    target
):

    with tf.GradientTape() as gen_tape, \
         tf.GradientTape() as disc_tape:


        # Generator creates fake image
        gen_output = generator(
            input_image,
            training=True
        )


        # Discriminator checks real image
        disc_real_output = discriminator(
            [
                input_image,
                target
            ],
            training=True
        )


        # Discriminator checks generated image
        disc_generated_output = discriminator(
            [
                input_image,
                gen_output
            ],
            training=True
        )


        # Calculate generator losses
        (
            gen_total_loss,
            gen_gan_loss,
            gen_l1_loss
        ) = generator_loss(
            disc_generated_output,
            gen_output,
            target
        )


        # Calculate discriminator loss
        disc_loss = discriminator_loss(
            disc_real_output,
            disc_generated_output
        )


    # Generator gradients
    generator_gradients = gen_tape.gradient(
        gen_total_loss,
        generator.trainable_variables
    )


    # Discriminator gradients
    discriminator_gradients = disc_tape.gradient(
        disc_loss,
        discriminator.trainable_variables
    )


    # Update Generator
    generator_optimizer.apply_gradients(
        zip(
            generator_gradients,
            generator.trainable_variables
        )
    )


    # Update Discriminator
    discriminator_optimizer.apply_gradients(
        zip(
            discriminator_gradients,
            discriminator.trainable_variables
        )
    )


    return (
        gen_total_loss,
        gen_gan_loss,
        gen_l1_loss,
        disc_loss
    )


# ------------------------------------------------------------
# 29. DISPLAY GENERATED RESULTS
# ------------------------------------------------------------

def generate_images(
    model,
    test_input,
    target,
    epoch
):

    prediction = model(
        test_input,
        training=False
    )


    plt.figure(
        figsize=(15, 5)
    )


    display_list = [
        test_input[0],
        target[0],
        prediction[0]
    ]


    titles = [
        "Input Sketch",
        "Real Image",
        "Generated Image"
    ]


    for i in range(3):

        plt.subplot(
            1,
            3,
            i + 1
        )


        plt.title(
            titles[i]
        )


        plt.imshow(
            (display_list[i] + 1) / 2
        )


        plt.axis("off")


    plt.suptitle(
        "Epoch " + str(epoch)
    )


    plt.show()


# ------------------------------------------------------------
# 30. PREPARE VALIDATION IMAGE
# ------------------------------------------------------------

test_dataset = tf.data.Dataset.from_tensor_slices(
    val_images
)


test_dataset = test_dataset.map(
    load_train_image,
    num_parallel_calls=tf.data.AUTOTUNE
)


test_dataset = test_dataset.batch(
    1
)


example_input, example_target = next(
    iter(test_dataset)
)


# ------------------------------------------------------------
# 31. TRAIN GAN
# ------------------------------------------------------------

print("\n======================================")
print("STARTING GAN TRAINING")
print("======================================")

for epoch in range(EPOCHS):

    start = time.time()


    print(
        "\nEpoch",
        epoch + 1,
        "/",
        EPOCHS
    )


    total_gen_loss = 0.0
    total_disc_loss = 0.0


    for step, (
        input_image,
        target
    ) in enumerate(train_dataset):


        (
            gen_loss,
            gen_gan_loss,
            gen_l1_loss,
            disc_loss
        ) = train_step(
            input_image,
            target
        )


        total_gen_loss += float(
            gen_loss
        )


        total_disc_loss += float(
            disc_loss
        )


        if step % 50 == 0:

            print(
                "Step:",
                step,
                "| Generator Loss:",
                round(
                    float(gen_loss),
                    4
                ),
                "| Discriminator Loss:",
                round(
                    float(disc_loss),
                    4
                )
            )


    # Average losses
    number_of_steps = len(
        train_images
    )


    print(
        "\nAverage Generator Loss:",
        total_gen_loss / number_of_steps
    )


    print(
        "Average Discriminator Loss:",
        total_disc_loss / number_of_steps
    )


    # Display result after every epoch
    generate_images(
        generator,
        example_input,
        example_target,
        epoch + 1
    )


    # Save checkpoint
    checkpoint.save(
        file_prefix=checkpoint_prefix
    )


    print(
        "Time taken:",
        round(
            time.time() - start,
            2
        ),
        "seconds"
    )


# ------------------------------------------------------------
# 32. FINAL RESULT
# ------------------------------------------------------------

print("\n======================================")
print("TRAINING COMPLETED")
print("======================================")


final_prediction = generator(
    example_input,
    training=False
)


plt.figure(
    figsize=(18, 6)
)


plt.subplot(1, 3, 1)

plt.imshow(
    (example_input[0] + 1) / 2
)

plt.title(
    "INPUT - SKETCH"
)

plt.axis("off")


plt.subplot(1, 3, 2)

plt.imshow(
    (example_target[0] + 1) / 2
)

plt.title(
    "ACTUAL REAL IMAGE"
)

plt.axis("off")


plt.subplot(1, 3, 3)

plt.imshow(
    (final_prediction[0] + 1) / 2
)

plt.title(
    "GAN GENERATED IMAGE"
)

plt.axis("off")


plt.show()


# ------------------------------------------------------------
# 33. SAVE TRAINED GENERATOR
# ------------------------------------------------------------

generator.save(
    "sketch_to_real_generator.keras"
)

print(
    "\nGenerator saved as:"
)

print(
    "sketch_to_real_generator.keras"
)


# ------------------------------------------------------------
# 34. SAVE DISCRIMINATOR
# ------------------------------------------------------------

discriminator.save(
    "sketch_to_real_discriminator.keras"
)

print(
    "Discriminator saved as:"
)

print(
    "sketch_to_real_discriminator.keras"
)


# ------------------------------------------------------------
# 35. DOWNLOAD GENERATOR
# ------------------------------------------------------------

from google.colab import files

print(
    "\nDownloading trained Generator..."
)

files.download(
    "sketch_to_real_generator.keras"
)


# ------------------------------------------------------------
# 36. FINAL MESSAGE
# ------------------------------------------------------------

print("\n======================================")
print("SKETCH-TO-REAL GAN PROJECT COMPLETED!")
print("======================================")

print("""
Project:
Sketch-to-Real Image Generation using Pix2Pix GAN

Architecture:
Generator  -> U-Net
Discriminator -> PatchGAN

Dataset:
CMP Facades

Input:
Building Sketch / Label

Output:
Generated Realistic Building Image
""")
```
