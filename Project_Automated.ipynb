```python
import requests
from pyngrok import ngrok

# Confirm that Streamlit is still running
response = requests.get(
    "http://127.0.0.1:8501/_stcore/health",
    timeout=10
)

if response.status_code == 200:
    print("Streamlit is healthy.")

    # Close any previous tunnels
    ngrok.kill()

    # Create a new public tunnel
    tunnel = ngrok.connect(8501, "http")

    print("\nOpen your dashboard using this NEW URL:")
    print(tunnel.public_url)
else:
    print("Streamlit is not responding. Restart the app first.")
```
